```python
# check duplicates of rows
duplicates_count = df.duplicated().sum()
print(f'⚠️ number of duplicates rows: {duplicates_count}') 
```

```python
# check duplicates of columns
t_duplicates = df.T.duplicated().sum()
print(f'⚠️ number of duplicates Columns: {t_duplicates}')
```

```python
df.drop_duplicates(inplace=True)
print(f'number of rows after dropping duplicates:{len(df)}')
```

```python
# change date into datetime
df['Date'] = pd.to_datetime(df['Date'])
print(df['Date'].dtype)
```

```python
df['Year'] = df['Date'].dt.year
df['Month_Name'] = df['Date'].dt.month_name()
```

```python
fig = px.box(
df,
y="Customer_Age",
title="Outlier (Customer_Age)"
)
fig.show()
```

```python
print('⚠️ Calculate Outliers')
col = df["Customer_Age"]
```

```python
Q1 = col.quantile(0.25)
print("Q1 =", Q1)
```

```python
Q3 = col.quantile(0.75)
print("Q3 =", Q3)
```

```python
IQR = Q3 - Q1
print("IQR =", IQR)
print('⚠️ Outlier Limits') 
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR
print("Lower Bound =", lower_bound)
print("Upper Bound =", upper_bound)
```

```python
df['Customer_Gender'] = df['Customer_Gender'].map({'M': 'Male', 'F': 'Female'})
```

# 🚲Bikes KPIs

```python
# top 3 countries
top_3_countries = df.groupby('Country')['Revenue'].sum().sort_values(ascending=False).head(3)
print("top 3 countries:")
print(top_3_countries)
```

```python
plt.figure(figsize=(14, 7))
top_3_countries = df.groupby('Country')['Revenue'].sum().sort_values(ascending=False).head(3)
sns.barplot(x=top_3_countries.index, y=top_3_countries.values, palette='viridis')
plt.title('Top 3 Countries by Revenue')
plt.ylabel('Total Revenue')
plt.tight_layout()
plt.show()
```

```python
# top product by quantity
top_category = df.groupby('Product_Category')['Order_Quantity'].sum().sort_values(ascending=False)
print("top product by quantity:")
print(top_category)
```

```python
plt.figure(figsize=(14, 7))
plt.subplot(1, 2, 2)
category_sales = df.groupby('Product_Category')['Order_Quantity'].sum().sort_values(ascending=False)
plt.pie(category_sales, labels=category_sales.index, autopct='%1.1f%%',  colors=['#ff9999','#66b3ff','#99ff99'])
plt.title('Sales Distribution by Category')
plt.tight_layout()
plt.show()
```

```python
#top 5 product by profit
plt.figure(figsize=(14, 7))
plt.subplot(1, 2, 2)
top_5_profit_products = df.groupby('Product')['Profit'].sum().sort_values(ascending=False).head(5)
sns.barplot(x=top_5_profit_products.values, y=top_5_profit_products.index, palette='magma')
plt.title('Top 5 Most Profit  Products')
plt.xlabel('Total Profit')
plt.tight_layout()
plt.show()
```

```python
#age distribution by sales
plt.figure(figsize=(12, 6))

sns.histplot(df['Customer_Age'], bins=30, kde=True, color='teal')

plt.axvline(df['Customer_Age'].mean(), color='red', linestyle='--', label=f'Mean: {df['Customer_Age'].mean():.1f}')
plt.axvline(df['Customer_Age'].median(), color='yellow', linestyle='-', label=f'Median: {df['Customer_Age'].median():.1f}')

plt.title('Customer Age Distribution (Who are our main buyers?)')
plt.xlabel('Age')
plt.ylabel('Frequency (Number of Sales)')
plt.legend()
plt.show()
```

```python
# Revenue by year
plt.figure(figsize=(16, 6))
yearly_revenue = df.groupby('Year')['Revenue'].sum()
yearly_revenue.plot(kind='line', marker='o', color='green', linewidth=2)
plt.title('Total Revenue Growth per Year')
plt.grid(True, linestyle='--', alpha=0.6)
plt.ylabel('Revenue')
plt.tight_layout
plt.show()
```

```python
# Revenue by month
plt.figure(figsize=(16, 6))
month_order = ['January', 'February', 'March', 'April', 'May', 'June', 
               'July', 'August', 'September', 'October', 'November', 'December']
monthly_sales = df.groupby('Month')['Revenue'].sum().reindex(month_order)
sns.barplot(x=monthly_sales.index, y=monthly_sales.values, palette='coolwarm')
plt.xticks(rotation=45)
plt.title('Total Revenue by Month')

plt.tight_layout()
plt.show()
```

```python
#age and gender by profit
age_gender_profit = df.groupby(['Age_Group', 'Customer_Gender'])['Profit'].mean().unstack()
age_gender_profit
```

```python
plt.figure(figsize=(12, 6))
sns.heatmap(age_gender_profit, annot=True, cmap='YlGnBu', fmt='.1f')
plt.title('Average Profit by Age Group and Gender')
plt.show()
```
